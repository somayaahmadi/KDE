```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("./data/heart_failure_clinical_records_dataset.csv")

df["ejection_fraction"].plot.kde()

plt.xlabel("Ejection Fraction")
plt.title("KDE of Ejection Fraction")
plt.show()
```

```python
for i in range(15):
    sample = df.sample(frac=1.0, replace=True)
    sample["ejection_fraction"].plot.kde()

plt.show()
```

4c. The original KDE appears more reliable around 20–30 and 45–55, where the resampled curves are relatively close together. There is noticeable variation around 35–40 and 55–65, where the resampled KDE curves are more spread out.
